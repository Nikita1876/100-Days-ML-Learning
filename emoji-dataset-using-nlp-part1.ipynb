```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "/kaggle/input/datasets/nikitagupta1812/emojidataset/train (1).txt",
    sep=";",
    header=None,
    names=["text", "emotion"]
)
```

## STEP-1 LOWERCASE ELEMENTS 

```python
df.isnull()
```

```python
unique_emotion = df['emotion'].unique()
emotion_numbers = {}
i=0
for emo in unique_emotion:
    emotion_numbers[emo] = i
    i+= 1


df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df
```

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

## STEP -2 REMOVING PUNCTUATIONS

```python
import string
def remove_punc(text):
    return text.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']= df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new+i
            return new
df['text'] = df['text'].apply(remove_punc)
```

## STEP -3 REMOVING STOPWORDS

```python
## CORPUS- Acollection of texts Like a dataset of Texts 
## SENTENCES - A GROUP PF WORDS THAT FORMS A MEANINGFUL STATEMENT
## TOKEN - EACH INDIVIDUAL WORD OR SYMBOL IN THE TEXT CALLED A TOKEN.
## 
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words) ## length of stopwords
```

```python
df.loc[1]['text']
```

```python
from nltk.tokenize import word_tokenize

def remove(text):
    words = word_tokenize(text)
    cleaned = []

    for i in words:
        if i not in stop_words:
            cleaned.append(i)

    return  ''.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
documents = [
    'i love Pizza',
    'Pizza Is Best',
    'i Love Pasta Also'
]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
print("Vocabulary: ", vectorizer.get_feature_names_out())
print("\nBow Matrix:\n ", X.toarray())
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test,y_train,y_test = train_test_split(df['text'],df['emotion'],test_size=0.20,random_state=42)
```

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)

```

```python
from sklearn.naive_bayes import MultinomialNB ## NB -NAIVE BAYES
from sklearn.metrics import accuracy_score
```

```python
nb_model.fit(X_train_bow,y_train)
```
